# 1. Why Continual World Models?

A long-lived agent does not solve one task once.

It encounters a **sequence of tasks**:

\[
\mathcal{T}_1 \rightarrow \mathcal{T}_2 \rightarrow \cdots \rightarrow \mathcal{T}_N
\]

The agent should ideally achieve **both**:

### Retention
Learn new tasks without destroying behavior on old tasks.

### Forward Transfer
Use previously learned knowledge to learn new tasks faster.

---

### Key distinction

| Setting | Main question |
|---|---|
| Multi-task RL | Can one model solve many tasks jointly? |
| Transfer RL | Can knowledge from old tasks help a new task? |
| Continual RL | Can the agent learn tasks sequentially while retaining and reusing knowledge? |

**Our interest is the third setting.**


# 2. Benchmark: Continual World

**Continual World** is a robotic continual-RL benchmark built on **Meta-World**.

### Core idea

Tasks are learned **sequentially**, rather than jointly.

A canonical sequence is:

\[
\text{Task}_1 \rightarrow \text{Task}_2 \rightarrow \cdots \rightarrow \text{Task}_{10}
\]

and CW20 repeats a 10-task sequence for a total of 20 task phases.

### Metrics that matter

\[
\boxed{\text{Final Performance}}
\]

\[
\boxed{\text{Forward Transfer}}
\]

\[
\boxed{\text{Forgetting}}
\]

A major point of the benchmark is that **low forgetting alone is not enough**.  
A useful continual learner should also produce **positive forward transfer**.

---

### Limitation for our future representation question

Continual World is useful for continual-learning evaluation, but it is primarily a **state-based robotic benchmark** rather than a visual representation-learning benchmark.

So it is a strong **baseline benchmark**, but may not be sufficient as the final testbed for a representation-focused project.

**Reference:** Wołczyk et al., *Continual World: A Robotic Benchmark for Continual Reinforcement Learning*, 2021.


# 3. Existing Continual World-Model Work

## Continual-Dreamer — CoLLAs 2023

**Question:** Are world models naturally useful for continual RL?

### Approach
- Dreamer-style latent world model
- selective experience replay
- task-agnostic continual learning
- evaluated on MiniGrid and MiniHack

### Main result
World models + replay can provide strong continual-learning performance.

### What it does *not* fully answer

It does not isolate a structured decomposition such as

\[
\text{shared physical dynamics}
+
\text{task-specific knowledge}
\]

and it is not focused on sequential robotic manipulation.

---

**Takeaway:**  
World models are already known to be useful for continual RL.  
A new project therefore needs a more specific mechanism than simply “use a world model for continual learning.”

**Reference:** Kessler et al., *The Effectiveness of World Models for Continual Reinforcement Learning*, CoLLAs 2023.


# 4. Life-Long World Model — ECML-PKDD 2025

## Continual Visual Reinforcement Learning with a Life-Long World Model

This work is especially close to our original direction.

### Problem
Learn visual dynamics across a sequence of non-stationary tasks **without forgetting previous dynamics**.

### Core idea
Treat the world model as a persistent simulator.

The method uses:
- task-specific latent dynamics modeling
- mixture-based latent structure
- generative experience replay
- imagined trajectories for policy optimization

### Conceptual assumption

\[
\text{Non-forgetting World Model}
\Rightarrow
\text{Reusable simulator for old + new tasks}
\]

---

### Why this matters for us

This paper already occupies part of the space:

\[
\text{continual learning}
+
\text{world-model retention}
+
\text{visual control}
\]

So our novelty cannot simply be:

> “Make the world model remember previous tasks.”

**Reference:** Pan et al., *Continual Visual Reinforcement Learning with a Life-Long World Model*, ECML-PKDD 2025.


# 5. Critical New Result

## The World Model Remembers, the Actor Forgets — 2026

This paper asks a more precise question:

> When Dreamer catastrophically forgets an old task, **which component actually forgot?**

### Experimental result

Under a setting where old experience remains available:

\[
\text{World Model Knowledge} \approx \text{retained}
\]

while

\[
\text{Actor Behavior} \rightarrow \text{collapses}
\]

The authors probe old-task information such as:
- reward discrimination
- value estimates
- termination structure

and find that much of this knowledge remains measurable in the world model.

### Intervention

With the world model frozen, lost skills can be recovered using **graded dream rehearsal / self-imitation** without new environment interaction.

---

## Interpretation

\[
\boxed{\text{Forgetting may be a behavior-readout problem, not purely a memory problem}}
\]

**Reference:** Nijjer, *The World Model Remembers, the Actor Forgets: Dream Rehearsal for Continual Model-Based RL*, 2026 preprint.


# 6. What Happens to Our Original Hypothesis?

### Original hypothesis

> A world model that separates **reusable physical dynamics** from **task-specific knowledge** can achieve both positive forward transfer and reduced catastrophic forgetting during sequential robotic skill acquisition.

This should now be split into **two different hypotheses**.

---

## Hypothesis A — Forward Transfer

\[
\text{Factorized / reusable dynamics representation}
\]

\[
\Downarrow
\]

\[
\text{better learning of future tasks}
\]

This remains plausible and well motivated.

---

## Hypothesis B — Reduced Behavioral Forgetting

\[
\text{Factorized world representation}
\]

\[
\stackrel{?}{\Longrightarrow}
\]

\[
\text{less forgetting of old behavior}
\]

This is **not automatically true**.

If the world model retains old-task knowledge while the actor forgets, then improving world-model memory alone may not solve behavioral forgetting.


# 7. Revised Research Question

The more interesting question may be:

> **What should a lifelong agent remember, where should it remember it, and how should reusable knowledge be converted back into behavior?**

This creates two coupled research directions.

---

## A. Representation / World-Model Side

\[
z =
[z_{\text{shared physics}},\;
z_{\text{task-specific}}]
\]

Questions:
- Which state variables should remain reusable across tasks?
- Can physical structure produce stronger forward transfer?
- What should be invariant vs task-conditioned?

---

## B. Behavior / Control Side

\[
\text{persistent world knowledge}
\rightarrow
\text{persistent or recoverable skills}
\]

Questions:
- Why can the actor fail even when the world model remembers?
- Should old skills be recovered by actor rehearsal?
- Could planning / MPC act as a more reliable readout?
- Can actor-critic and planning be combined?


# 8. Connection to the Current Physical-Factor Project

The current project asks:

> Under a fixed data and model budget, when does explicitly exposing a physical factor improve compact action-conditioned dynamics learning?

That can naturally connect to continual world models later.

### Possible progression

\[
\textbf{Explicit Physical Factor}
\]

\[
\Downarrow
\]

\[
\textbf{Better Structured Dynamics Representation}
\]

\[
\Downarrow
\]

\[
\textbf{Multi-Task Dynamics Reuse}
\]

\[
\Downarrow
\]

\[
\textbf{Continual Robotic Skill Learning}
\]

So the current project can potentially provide evidence about **which physical information is worth preserving explicitly** before we study continual reuse.

---

### Longer-term idea

Instead of asking only

> “Does factorization reduce forgetting?”

ask

> “Which physical representations remain reusable across changing tasks, and how should an agent exploit them for transfer and skill recovery?”
